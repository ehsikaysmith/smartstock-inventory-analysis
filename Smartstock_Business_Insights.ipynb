# SmartStock: Inventory and Sales Analysis

## Project Overview

SmartStock is a data analytics project designed to help a business understand its sales performance, profitability, and inventory levels.

The project uses Python and pandas to clean and analyse sales, product, and inventory data. It also identifies profitable products and items that require replenishment.

## Business Objectives

- Analyse sales revenue, costs, and profitability.
- Identify the most profitable products.
- Evaluate inventory levels and identify products requiring replenishment.
- Generate actionable business recommendations.
- Present findings through clear tables and visualizations.

## Tools and Technologies

- Python
- Jupyter Notebook
- pandas
- Matplotlib
- Data cleaning and exploratory data analysis

## Expected Business Value

The analysis aims to support better purchasing decisions, improve inventory management, and help the business understand which products contribute most to its profitability.

- Total sales revenue: GH₵639,284.34.

**Business insight:** The analysis indicates that the business generated a positive profit from its recorded sales. Monitoring profit margins can help management evaluate pricing decisions and control product costs.

## Recommendations

1. Monitor the profitability of leading products regularly.
2. Replenish the products identified in the replenishment report.
3. Investigate missing or invalid inventory data before making purchasing decisions.
4. Review sales, costs, and profit margins periodically.
5. Use the project's visualizations and reports to support evidence-based business decisions.

## Methodology

### 1. Data Loading
Loaded the sales, product, and inventory datasets into Python using pandas.

### 2. Data Quality Assessment
Inspected the datasets for missing values, duplicate records, invalid quantities, and potential data quality issues.

### 3. Data Cleaning
Removed duplicate sales records and investigated invalid sales quantities. Reviewed missing inventory values to identify limitations that could affect the analysis.

### 4. Sales and Profitability Analysis
Calculated sales revenue, cost of goods sold, profit, and profit margins. Aggregated results by product to identify the leading contributors to profit.

### 5. Inventory Analysis
Examined stock levels against reorder levels to identify products requiring replenishment.

### 6. Reporting and Visualization
Created charts and summary tables to communicate the results clearly. Exported the replenishment report to CSV for practical business use.

## Project Limitations

- Missing inventory values may affect the accuracy of some inventory summaries.
- Historical sales performance does not guarantee future product profitability.
- Replenishment recommendations depend on the accuracy of the recorded stock levels and reorder thresholds.

## Conclusion

SmartStock demonstrates how Python-based data analysis can turn raw sales and inventory records into actionable business insights. The project combines data cleaning, exploratory analysis, profitability calculations, visualization, and report generation to support better business decisions.

```python
total_revenue = 639548.57
total_profit = 190205.98

profit_margin = (total_profit / total_revenue) * 100

print(f"Total revenue: GHS{total_revenue:,.2f}")
print(f"Total profit: GHS{total_profit:,.2f}")
print(f"Profit margin: {profit_margin:.2f}%")
```
